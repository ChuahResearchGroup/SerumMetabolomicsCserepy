```python
import pandas as pd
```

```python
adni = pd.read_csv("ADNIMERGE.csv", low_memory=False)
```

```python
adni = adni[['RID','VISCODE','DX_bl','DX','Month']]
```

```python
adni['DX_bl'] = adni['DX_bl'].replace({'AD': 'Dementia'})
```

```python
#dropping rows where DX is missing
adni = adni.dropna(subset=['DX'])
```

```python
#drop SMC in adni
adni = adni[~adni["DX"].isin(["SMC"]) & ~adni["DX_bl"].isin(["SMC"])]
```

```python
adni.to_csv("adniclean.csv", index=False)
```
